```python
import pandas as pd
import pingouin as pg
```

```python
data = pd.DataFrame({
    "Participant": [1, 2, 3, 4, 5, 6],
    "RT_before": [510, 480, 530, 495, 505, 520],
    "RT_after":  [490, 485, 500, 480, 500, 510],
})

data["Difference"] = data["RT_before"] - data["RT_after"]
data
```

```python
result = pg.ttest(
    data["RT_before"],
    data["RT_after"],
    paired=True,
    alternative="two-sided",
)

result.round(4)
```

```python
pg.ttest(
    data["RT_before"],
    data["RT_after"],
    paired=True,
    alternative="greater",
)
```
