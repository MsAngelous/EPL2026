Now, create your own dicionary below. Name it 'data'. Apply the patterns you noticed above when creating your own dataframe. The dataframe should have 6 columns: words, languages, parts_of_speech, frequencies, lengths, and len_above_6. How do you make a dictionary into a dataframe again?

Hint: use pd.DataFrame()

```python
def pos_identifier(list_of_pos):
    nouns = 0
    verbs = 0
    adjectives = 0

    for pos in list_of_pos:
        if pos == "noun":
            nouns += 1
        elif pos == "verb":
            verbs += 1
        elif pos == "adj":
            adjectives += 1

    print(f"Nouns: {nouns}, verbs: {verbs}, adjectives: {adjectives}")


pos_identifier(data["parts_of_speech"])
```

6. Now that we have our dataframe, let's take a look at it: use data.head() to see the first 5 rows.

```python
def categorize_frequency(frequency):
    if frequency >= 65:
        return "very frequent"
    if frequency > 40:
        return "moderately frequent"
    return "not frequent"


data["frequency_category"] = data["frequencies"].apply(categorize_frequency)
data[["words", "frequencies", "frequency_category"]]
```
