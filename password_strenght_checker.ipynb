```python
import string
from getpass import getpass

password = getpass("Enter a made-up password to check: ")

long_enough = len(password) >= 12
has_uppercase = any(character.isupper() for character in password)
has_lowercase = any(character.islower() for character in password)
has_number = any(character.isdigit() for character in password)
has_special = any(character in string.punctuation for character in password)

if (
    long_enough
    and has_uppercase
    and has_lowercase
    and has_number
    and has_special
):
    print("Strong password according to this exercise's rules.")
else:
    print("Weak password according to this exercise's rules.")

    if not long_enough:
        print("- Use at least 12 characters.")
    if not has_uppercase:
        print("- Add an uppercase letter.")
    if not has_lowercase:
        print("- Add a lowercase letter.")
    if not has_number:
        print("- Add a number.")
    if not has_special:
        print("- Add a special character, such as !, @, or #.")
```
