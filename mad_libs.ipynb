```python
noun = input("Enter a noun, like 'penguin': ")
verb = input("Enter a verb, like 'dance': ")
adjective = input("Enter an adjective, like 'grumpy': ")
food = input("Enter a food, like 'pancakes': ")

story = (
    f"Yesterday, a {adjective} {noun} walked into my classroom. "
    f"It tried to {verb} on the teacher's desk. "
    f"The teacher gave it {food} and asked it to come back after lunch."
)

print()
print(story)
```
